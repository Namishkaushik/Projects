### Step – 6 : Exploratory Data Analysis (EDA)

This section explores key questions to identify relationships between student performance and various demographic, academic, and social factors.

#### Q.1. Average final score (G3) by gender

```python
gender_avg = df.groupby('sex')['G3'].mean()
print(gender_avg)

plt.figure(figsize=(6, 4))
ax = sns.barplot(x=gender_avg.index, y=gender_avg.values, hue=gender_avg.index, palette='pastel', legend=False)
plt.title('Average Final Grade (G3) by Gender')
plt.xlabel('Gender')
plt.ylabel('Average G3')
for bar in ax.patches:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            f'{bar.get_height():.2f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('Q1_gender_vs_g3.png')
plt.show()
```

Female students have a slightly higher average final score (G3) compared to male students. This indicates a modest difference in academic performance by gender.

#### Q.2. Does study time affect final grades?

```python
studytime_map = {1: '< 2 hours', 2: '2-5 hours', 3: '5-10 hours', 4: '> 10 hours'}
order = ['< 2 hours', '2-5 hours', '5-10 hours', '> 10 hours']
study_avg = df.assign(studytime_label=df['studytime'].map(studytime_map)).groupby('studytime_label')['G3'].mean().loc[order]

plt.figure(figsize=(7, 4))
ax = sns.barplot(x=study_avg.index, y=study_avg.values, hue=study_avg.index, palette='Blues_d', legend=False)
plt.title('Average Final Grade (G3) by Study Time')
plt.xlabel('Study Time per Week')
plt.ylabel('Average G3')
for bar in ax.patches:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            f'{bar.get_height():.2f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('Q2_studytime_vs_g3.png')
plt.show()
```

Students with higher study time tend to have higher average final grades, suggesting a positive association between time spent studying and academic performance. Although the increase is not perfectly linear, higher study time categories generally show better average grades.

#### Q.3. Do students with internet access perform better?

```python
internet_avg = df.groupby('internet')['G3'].mean()
print(internet_avg)

plt.figure(figsize=(6, 4))
ax = sns.barplot(x=internet_avg.index, y=internet_avg.values, hue=internet_avg.index, palette='Set2', legend=False)
plt.title('Average Final Grade (G3) by Internet Access')
plt.xlabel('Internet Access')
plt.ylabel('Average G3')
for bar in ax.patches:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            f'{bar.get_height():.2f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('Q3_internet_vs_g3.png')
plt.show()
```

Students with internet access tend to have higher average final grades compared to those without. This indicates a positive association between internet access and academic performance.

#### Q.4. What is the impact of parental education on grades?

```python
fedu_avg = df.groupby('Fedu')['G3'].mean()
medu_avg = df.groupby('Medu')['G3'].mean()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
edu_labels = {0: 'None', 1: 'Primary', 2: 'Middle', 3: 'Secondary', 4: 'Higher'}

axes[0].bar(fedu_avg.index.map(edu_labels), fedu_avg.values, color='salmon')
axes[0].set_title("Father's Education vs Avg G3")
axes[0].set_xlabel("Father's Education Level")
axes[0].set_ylabel('Average G3')
for i, v in enumerate(fedu_avg.values):
    axes[0].text(i, v + 0.1, f'{v:.2f}', ha='center', fontsize=9)

axes[1].bar(medu_avg.index.map(edu_labels), medu_avg.values, color='mediumpurple')
axes[1].set_title("Mother's Education vs Avg G3")
axes[1].set_xlabel("Mother's Education Level")
axes[1].set_ylabel('Average G3')
for i, v in enumerate(medu_avg.values):
    axes[1].text(i, v + 0.1, f'{v:.2f}', ha='center', fontsize=9)

plt.tight_layout()
plt.savefig('Q4_parental_edu_vs_g3.png')
plt.show()
```

Student final grades show a positive association with both parents' education levels. The association appears slightly stronger for mother's education — higher maternal education levels correspond to higher average final grades compared to paternal education.

#### Q.5. What is the relationship between student absences and academic performance?

```python
bins = [0, 5, 10, 20, 40]
labels = ['0-5', '5-10', '10-20', '20-40']
absence_data = df.assign(absences_label=pd.cut(df['absences'], bins=bins, labels=labels))     .groupby('absences_label', observed=False)['G3'].agg(['count', 'mean'])
print(absence_data)

plt.figure(figsize=(7, 4))
ax = sns.barplot(x=absence_data.index, y=absence_data['mean'], hue=absence_data.index, palette='Oranges_r', legend=False)
plt.title('Average Final Grade (G3) by Absence Range')
plt.xlabel('Number of Absences')
plt.ylabel('Average G3')
for bar in ax.patches:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            f'{bar.get_height():.2f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('Q5_absences_vs_g3.png')
plt.show()
```

The analysis shows a negative relationship between student absences and final grades. Students in the lowest absence range (0–5) have the highest average final grades. The 20–40 range shows a slight increase, but this category contains very few students (n=8) and should be interpreted with caution.

#### Q.6. How does alcohol consumption relate to students' grades?

```python
alc_map = {1: 'Very Low', 2: 'Low', 3: 'Moderate', 4: 'High', 5: 'Very High'}
order = ['Very Low', 'Low', 'Moderate', 'High', 'Very High']

dalc_data = df.assign(Dalc_label=df['Dalc'].map(alc_map)).groupby('Dalc_label')['G3'].agg(['count', 'mean']).loc[order]
walc_data = df.assign(Walc_label=df['Walc'].map(alc_map)).groupby('Walc_label')['G3'].agg(['count', 'mean']).loc[order]

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].bar(dalc_data.index, dalc_data['mean'], color='tomato')
axes[0].set_title('Weekday Alcohol (Dalc) vs Avg G3')
axes[0].set_xlabel('Alcohol Consumption Level')
axes[0].set_ylabel('Average G3')
axes[0].tick_params(axis='x', rotation=15)
for i, v in enumerate(dalc_data['mean'].values):
    axes[0].text(i, v + 0.1, f'{v:.2f}', ha='center', fontsize=8)

axes[1].bar(walc_data.index, walc_data['mean'], color='coral')
axes[1].set_title('Weekend Alcohol (Walc) vs Avg G3')
axes[1].set_xlabel('Alcohol Consumption Level')
axes[1].set_ylabel('Average G3')
axes[1].tick_params(axis='x', rotation=15)
for i, v in enumerate(walc_data['mean'].values):
    axes[1].text(i, v + 0.1, f'{v:.2f}', ha='center', fontsize=8)

plt.tight_layout()
plt.savefig('Q6_alcohol_vs_g3.png')
plt.show()
```

Higher alcohol consumption is associated with lower academic performance. The trend is clearer for weekend alcohol use (Walc), which shows a more consistent decline. Higher consumption categories contain fewer students, so results should be interpreted cautiously.

#### Q.7. Is there a difference in academic performance across schools?

```python
school_data = df.groupby('school')['G3'].agg(['count', 'mean'])
print(school_data)

plt.figure(figsize=(6, 4))
ax = sns.barplot(x=school_data.index, y=school_data['mean'], hue=school_data.index, palette='muted', legend=False)
plt.title('Average Final Grade (G3) by School')
plt.xlabel('School')
plt.ylabel('Average G3')
for bar in ax.patches:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            f'{bar.get_height():.2f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('Q7_school_vs_g3.png')
plt.show()
```

Students from school GP have a higher average final grade compared to students from school MS. This suggests that school-level factors — such as teaching quality, resources, or environment — may influence student performance.

#### Q.8. Does family support influence student performance?

```python
famsup_data = df.groupby('famsup')['G3'].agg(['count', 'mean'])
print(famsup_data)

plt.figure(figsize=(6, 4))
ax = sns.barplot(x=famsup_data.index, y=famsup_data['mean'], hue=famsup_data.index, palette='Set3', legend=False)
plt.title('Average Final Grade (G3) by Family Support')
plt.xlabel('Family Support')
plt.ylabel('Average G3')
for bar in ax.patches:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            f'{bar.get_height():.2f}', ha='center', fontsize=10)
plt.tight_layout()
plt.savefig('Q8_famsup_vs_g3.png')
plt.show()
```

Students with family support score slightly higher on average, suggesting a modest positive association between family support and final grades.

### Step - 7 : Correlation Heatmap

```python
numeric_cols = ['studytime', 'failures', 'absences', 'Dalc', 'Walc',
                'Medu', 'Fedu', 'famrel', 'goout', 'G1', 'G2', 'G3']

plt.figure(figsize=(11, 8))
corr = df[numeric_cols].corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, linewidths=0.5, annot_kws={'size': 9})
plt.title('Correlation Heatmap of Numeric Features', fontsize=13)
plt.tight_layout()
plt.savefig('correlation_heatmap.png')
plt.show()
```

The heatmap reveals the following key correlations with G3 (final grade):
- **G1 and G2** have the strongest positive correlation with G3 — prior grades are the best predictor of final performance.
- **failures** shows a strong negative correlation with G3 — students with more past failures tend to score lower.
- **studytime** has a modest positive correlation with G3.
- **absences**, **Dalc**, and **Walc** show weak negative correlations with G3.
- **Medu** and **Fedu** show small positive correlations with G3.
